#The Transformation Logic

```python
query = """
SELECT
    ROW_NUMBER () OVER (ORDER BY p.product_number, p.start_date) AS product_key, --Surrogate Key
    p.product_id,
    p.product_number,
    p.product_name,
    p.product_cost,
    p.product_line,
    p.start_date,
    p.category_id,
    pc.category,
    pc.subcategory
FROM silver.crm_products p
LEFT JOIN workspace.silver.erp_product_category pc
ON p.category_id = pc.category_id
WHERE p.end_date IS NULL; -- Filtering out all historical data
"""
df = spark.sql(query)
```

```python
df.limit(10).display()
```

#Writing Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_product")
```
