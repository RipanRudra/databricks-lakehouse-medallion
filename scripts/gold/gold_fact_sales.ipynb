#The Transformation Logic

```python
query = """
SELECT
  s.order_number,
  p.product_key,
  c.customer_key,
  s.order_date,
  s.ship_date,
  s.due_date,
  s.sales_amount,
  s.quantity,
  s.price
FROM workspace.silver.crm_sales s
LEFT JOIN workspace.gold.dim_product p ON s.product_number = P.product_number
LEFT JOIN workspace.gold.dim_customer c ON s.customer_id = C.customer_id
"""
df = spark.sql(query)
display(df)
```

#Write in Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
